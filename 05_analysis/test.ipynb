```python
import pandas as pd 

fn = '/projects/p32903/Alex2/datasets/STU00225089/.clincal_notes/cgi_batch_agg_2_clean.xlsx'
sheet_name = 'stimulus'

df = pd.read_excel(fn, sheet_name=sheet_name)

course_placements = (
    df.groupby(["subjid", "series_num"], as_index=False)["electrode_placement"]
      .agg(lambda values: values.mode().iat[0])
      .rename(columns={"electrode_placement": "modal_electrode_placement"})
)
```
